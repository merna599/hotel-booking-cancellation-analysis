```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create a separate dataset for cleaning
df_clean = df.copy()
rows_before = len(df_clean)

# Remove identical rows
df_clean = df_clean.drop_duplicates()

# Remove records with an unknown number of children
df_clean = df_clean.dropna(subset=["children"])

# Fill missing country values
df_clean["country"] = df_clean["country"].fillna("Unknown")

# Create useful analysis columns
df_clean["total_guests"] = (
    df_clean["adults"]
    + df_clean["children"]
    + df_clean["babies"]
)

df_clean["total_nights"] = (
    df_clean["stays_in_weekend_nights"]
    + df_clean["stays_in_week_nights"]
)

# Keep records that meet the analysis rules
df_clean = df_clean.loc[
    (df_clean["adr"] >= 0)
    & (df_clean["total_guests"] > 0)
    & (df_clean["total_nights"] > 0)
].copy()

# Use whole numbers for guest counts
df_clean["children"] = df_clean["children"].astype(int)
df_clean["total_guests"] = df_clean["total_guests"].astype(int)

df_clean = df_clean.reset_index(drop=True)

print(f"Rows before cleaning: {rows_before:,}")
print(f"Rows after cleaning: {len(df_clean):,}")
print(f"Rows removed: {rows_before - len(df_clean):,}")
```

```python
df_clean.to_csv("hotel_bookings_cleaned.csv", index=False)
```

```python
month_numbers = {
    "January": 1, "February": 2, "March": 3,
    "April": 4, "May": 5, "June": 6,
    "July": 7, "August": 8, "September": 9,
    "October": 10, "November": 11, "December": 12
}

df_clean["arrival_date"] = pd.to_datetime(
    {
        "year": df_clean["arrival_date_year"],
        "month": df_clean["arrival_date_month"].map(month_numbers),
        "day": df_clean["arrival_date_day_of_month"]
    },
    errors="coerce"
)
```

```python
hotel_summary = df_clean.groupby("hotel").agg(
    total_bookings=("is_canceled", "size"),
    canceled_bookings=("is_canceled", "sum"),
    cancellation_rate=("is_canceled", "mean")
)

hotel_summary["cancellation_rate"] = (
    hotel_summary["cancellation_rate"] * 100
).round(2)

hotel_summary
```

```python
ax = hotel_summary["cancellation_rate"].plot(
    kind="bar",
    figsize=(7, 4),
    color=["#167D9A", "#67BDC6"],
    rot=0
)

ax.set_title("Cancellation Rate by Hotel Type")
ax.set_xlabel("Hotel Type")
ax.set_ylabel("Cancellation Rate (%)")
ax.set_ylim(0, 100)

for container in ax.containers:
    ax.bar_label(container, fmt="%.2f%%")

plt.tight_layout()
plt.show()
```

City Hotel had a cancellation rate of 30.20%, compared with
23.71% for Resort Hotel, a difference of 6.49 percentage points
in the cleaned dataset.

```python
feature_columns = [
    "hotel",
    "lead_time",
    "arrival_date_month",
    "adults",
    "children",
    "babies",
    "meal",
    "country",
    "market_segment",
    "distribution_channel",
    "is_repeated_guest",
    "reserved_room_type",
    "customer_type",
    "total_nights"
]

X = df_clean[feature_columns].copy()
y = df_clean["is_canceled"].copy()
```

```python
arrival_dates = pd.to_datetime(df_clean["arrival_date"])

train_mask = arrival_dates < "2017-01-01"

validation_mask = (
    (arrival_dates >= "2017-01-01")
    & (arrival_dates < "2017-05-01")
)

test_mask = arrival_dates >= "2017-05-01"

X_train = X.loc[train_mask].copy()
y_train = y.loc[train_mask].copy()

X_val = X.loc[validation_mask].copy()
y_val = y.loc[validation_mask].copy()

X_test = X.loc[test_mask].copy()
y_test = y.loc[test_mask].copy()

print("Training rows:", len(X_train))
print("Validation rows:", len(X_val))
print("Test rows:", len(X_test))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

numeric_columns = X_train.select_dtypes(
    include="number"
).columns.tolist()

categorical_columns = X_train.select_dtypes(
    exclude="number"
).columns.tolist()

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", StandardScaler(), numeric_columns),
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_columns
        )
    ]
)
```

```python
X_train_prepared = preprocessor.fit_transform(X_train)

X_val_prepared = preprocessor.transform(X_val)
X_test_prepared = preprocessor.transform(X_test)

print("Training shape:", X_train_prepared.shape)
print("Validation shape:", X_val_prepared.shape)
print("Test shape:", X_test_prepared.shape)
```

```python
from sklearn.linear_model import LogisticRegression

lr_model = LogisticRegression(
    max_iter=2000,
    random_state=42
)

lr_model.fit(X_train_prepared, y_train)
```

```python
val_predictions = lr_model.predict(X_val_prepared)

val_probabilities = lr_model.predict_proba(
    X_val_prepared
)[:, 1]
```

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)

baseline = DummyClassifier(strategy="most_frequent")
baseline.fit(X_train_prepared, y_train)

baseline_predictions = baseline.predict(X_val_prepared)

print(
    "Baseline accuracy:",
    round(accuracy_score(y_val, baseline_predictions), 3)
)

print(
    "Logistic Regression accuracy:",
    round(accuracy_score(y_val, val_predictions), 3)
)

print(
    "Precision:",
    round(precision_score(y_val, val_predictions, zero_division=0), 3)
)

print(
    "Recall:",
    round(recall_score(y_val, val_predictions, zero_division=0), 3)
)

print(
    "F1 score:",
    round(f1_score(y_val, val_predictions, zero_division=0), 3)
)

print(
    "ROC AUC:",
    round(roc_auc_score(y_val, val_probabilities), 3)
)

print(
    "Average precision:",
    round(average_precision_score(y_val, val_probabilities), 3)
)

print("Validation cancellation rate:", round(y_val.mean(), 3))
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    y_val,
    val_predictions,
    labels=[0, 1],
    display_labels=["Not Canceled", "Canceled"],
    cmap="Blues",
    values_format="d"
)

plt.title("Logistic Regression — Validation Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
lr_balanced = LogisticRegression(
    max_iter=2000,
    class_weight="balanced",
    random_state=42
)

lr_balanced.fit(X_train_prepared, y_train)

balanced_predictions = lr_balanced.predict(X_val_prepared)

print(
    "Precision:",
    round(precision_score(
        y_val, balanced_predictions, zero_division=0
    ), 3)
)

print(
    "Recall:",
    round(recall_score(
        y_val, balanced_predictions, zero_division=0
    ), 3)
)

print(
    "F1 score:",
    round(f1_score(
        y_val, balanced_predictions, zero_division=0
    ), 3)
)
```

```python
comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Balanced Logistic Regression"
    ],
    "Precision": [
        precision_score(y_val, val_predictions, zero_division=0),
        precision_score(y_val, balanced_predictions, zero_division=0)
    ],
    "Recall": [
        recall_score(y_val, val_predictions, zero_division=0),
        recall_score(y_val, balanced_predictions, zero_division=0)
    ],
    "F1": [
        f1_score(y_val, val_predictions, zero_division=0),
        f1_score(y_val, balanced_predictions, zero_division=0)
    ]
})

comparison.round(3)
```

Class weighting increased validation recall from 25.1% to
62.3% and F1 from 36.1% to 54.6%, while precision decreased
from 64.4% to 48.5%. The balanced model detected more
cancellations but generated more false alarms.

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=15,
    min_samples_leaf=5,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train_prepared, y_train)
```

```python
rf_predictions = rf_model.predict(X_val_prepared)

rf_probabilities = rf_model.predict_proba(
    X_val_prepared
)[:, 1]
```

```python
rf_result = pd.DataFrame({
    "Model": ["Random Forest"],
    "Precision": [
        precision_score(y_val, rf_predictions, zero_division=0)
    ],
    "Recall": [
        recall_score(y_val, rf_predictions, zero_division=0)
    ],
    "F1": [
        f1_score(y_val, rf_predictions, zero_division=0)
    ]
})

comparison = pd.concat(
    [
        comparison[comparison["Model"] != "Random Forest"],
        rf_result
    ],
    ignore_index=True
)

comparison.round(3)
```

Random Forest achieved the highest validation F1 score (0.590)
and recall (0.777) among the models evaluated so far.
Its precision was 0.476, indicating a trade-off between
detecting more cancellations and generating false alarms.

```python
import tensorflow as tf
from scipy import sparse

tf.keras.utils.set_random_seed(42)

def to_dense_float32(data):
    if sparse.issparse(data):
        return data.toarray().astype("float32")
    return np.asarray(data, dtype="float32")

X_train_nn = to_dense_float32(X_train_prepared)
X_val_nn = to_dense_float32(X_val_prepared)
```

```python
nn_model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(X_train_nn.shape[1],)),
    tf.keras.layers.Dense(64, activation="relu"),
    tf.keras.layers.Dropout(0.2),
    tf.keras.layers.Dense(32, activation="relu"),
    tf.keras.layers.Dense(1, activation="sigmoid")
])

nn_model.compile(
    optimizer="adam",
    loss="binary_crossentropy"
)
```

```python
from sklearn.utils.class_weight import compute_class_weight

classes = np.unique(y_train)

weights = compute_class_weight(
    class_weight="balanced",
    classes=classes,
    y=y_train.to_numpy()
)

class_weights = dict(zip(classes.tolist(), weights.tolist()))

early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

history = nn_model.fit(
    X_train_nn,
    y_train.to_numpy(),
    validation_data=(X_val_nn, y_val.to_numpy()),
    epochs=30,
    batch_size=256,
    class_weight=class_weights,
    callbacks=[early_stopping],
    verbose=1
)
```

```python
nn_probabilities = nn_model.predict(
    X_val_nn,
    verbose=0
).ravel()

nn_predictions = (nn_probabilities >= 0.5).astype(int)

nn_result = pd.DataFrame({
    "Model": ["Neural Network"],
    "Precision": [
        precision_score(y_val, nn_predictions, zero_division=0)
    ],
    "Recall": [
        recall_score(y_val, nn_predictions, zero_division=0)
    ],
    "F1": [
        f1_score(y_val, nn_predictions, zero_division=0)
    ]
})

comparison = pd.concat(
    [
        comparison[comparison["Model"] != "Neural Network"],
        nn_result
    ],
    ignore_index=True
)

comparison.round(3)
```

```python
test_probabilities = rf_model.predict_proba(
    X_test_prepared
)[:, 1]

test_predictions = (test_probabilities >= 0.5).astype(int)

test_results = pd.DataFrame({
    "Model": ["Random Forest"],
    "Dataset": ["Test"],
    "Accuracy": [
        accuracy_score(y_test, test_predictions)
    ],
    "Precision": [
        precision_score(y_test, test_predictions, zero_division=0)
    ],
    "Recall": [
        recall_score(y_test, test_predictions, zero_division=0)
    ],
    "F1": [
        f1_score(y_test, test_predictions, zero_division=0)
    ],
    "ROC_AUC": [
        roc_auc_score(y_test, test_probabilities)
    ],
    "Average_Precision": [
        average_precision_score(y_test, test_probabilities)
    ]
})

test_results.round(3)
```

```python
powerbi_predictions = df_clean.loc[
    X_test.index,
    [
        "hotel",
        "arrival_date",
        "market_segment",
        "lead_time",
        "is_repeated_guest",
        "total_nights",
        "is_canceled"
    ]
].copy()

powerbi_predictions["predicted_cancellation"] = test_predictions
powerbi_predictions["model_score"] = test_probabilities

powerbi_predictions["prediction_correct"] = (
    powerbi_predictions["is_canceled"]
    == powerbi_predictions["predicted_cancellation"]
)

powerbi_predictions.to_csv(
    "hotel_test_predictions.csv",
    index=False
)

comparison.to_csv(
    "model_validation_comparison.csv",
    index=False
)

test_results.to_csv(
    "model_test_results.csv",
    index=False
)
```
